```python
import kagglehub
import shutil
import os
import pandas as pd
import warnings 
warnings.filterwarnings("ignore")

# Define the desired download location
custom_path = "D:\SAMITH\Tableau-Projects\Logistics and Supply Chain Dashboard"

# Download the dataset (this will be stored in the default cache location)
path = kagglehub.dataset_download("datasetengineer/logistics-and-supply-chain-dataset")

# Find the versioned folder inside the downloaded dataset
versioned_folder = os.path.join(path, "versions")
if os.path.exists(versioned_folder):
    subfolders = os.listdir(versioned_folder)
    if subfolders:
        dataset_path = os.path.join(versioned_folder, subfolders[0])  # Get the first version folder
    else:
        dataset_path = path
else:
    dataset_path = path

# Ensure the destination directory exists
os.makedirs(custom_path, exist_ok=True)

# Move dataset contents instead of the whole "1" folder
for item in os.listdir(dataset_path):
    shutil.move(os.path.join(dataset_path, item), custom_path)

print("Dataset moved to:", custom_path)
```
